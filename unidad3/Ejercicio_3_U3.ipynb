```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
X=[0,1,2,3,4,5] #Valores posibles de la variable aleatoria
p=[0.05,0.10,0.25,0.30,0.20,0.10] #Probabilidades correspondientes

#Inicializar una variable para acumular la suma
valor_esperado=0

#Recorrer los valores y sus probabilidades
for i in range(len(X)): # Corrected indentation and used X instead of x
  xi=X[i]   #Valor de la variable aleatoria
  pi=p[i]   #Probabilidad de ese valor
  producto = xi*pi     #multiplicar valor por su probabilidad
  valor_esperado += producto #sumar al acumulador - Corrected line
  #Mostrar cada paso
  print(f"x={xi},p={pi},x*p={producto:.2f}")

#Mostrar resultado final
print(f"\nValor esperado (Esperanza): {valor_esperado:.2f}") # Corrected indentation
```

```python
x=[0,1,2,3,4,5]
p=[0.05,0.10,0.25,0.30,0.20,0.10]

#Calcular E(X)
esperanza=0
for i in range(len(x)):
  esperanza+=x[i]*p[i]

#Calcular E(X^2)
esperanza_cuadrado=0
for i in range(len(x)):
  esperanza_cuadrado+= (x[i]**2)*p[i]

#Calcular varianza y desviación estándar
varianza = esperanza_cuadrado-esperanza**2
desviacion= math.sqrt(varianza)

#Mostrar resultados
print(f"E(X)={esperanza:.2f}")
print(f"E(X^2)={esperanza_cuadrado:.2f}")
print(f"Varianza={varianza:.2f}")
print(f"Desviación estándar={desviacion:.2f}")
```

```python
x=[0,1,2,3,4,5] #Valores posibles de la variable aleatoria
p=[0.05,0.10,0.25,0.30,0.20,0.10] #Probabilidades correspondientes

#Calcular función acumulada
F=np.cumsum(p)

#Mostrar resultados
for xi,fi in zip(x,F):
  print(f"F({xi})={fi:.2f}")

#Graficar
plt.step(x,F,where='post')
plt.title("Función de Distribución Acumulada (FDA)")
plt.xlabel("Calificación")
plt.ylabel("Probabilidad")
plt.grid(True)
plt.ylim(0,1.05)
plt.show()
```
