##Distribución Poisson y Binomial

#Problema 1


#La distribución de Poisson es la adecuada porque el problema describe el conteo de eventos (ingresos a la unidad de cuidados intensivos) que ocurren de manera independiente en un intervalo de tiempo continuo.

```python
from scipy.stats import poisson

# Parámetros de la distribución de Poisson
media_pacientes = 5

# a) Número exacto de personas ingresadas
x = 2
probabilidad_2_personas = poisson.pmf(x, media_pacientes)

# b) Probabilidad de que exceda de 10
probabilidad_exceda_10 = 1 - poisson.cdf(10, media_pacientes)

print("La probabilidad de que ingresen exactamente 2 personas es:", probabilidad_2_personas)
print("La probabilidad de que ingresen más de 10 personas es:", probabilidad_exceda_10)
```

#Problema 2

#La solución es de tipo binomial, porque aquí hay un número fijo de motores en cada embarque y cada motor solo puede clasificarse como defectuoso o no defectuoso, y la probabilidad de defecto es constante.

```python
from scipy.stats import binom

# Parámetros de la distribución binomial
n = 1000
p = 0.001

# Calcular probabilidades
probabilidad_x_0 = binom.pmf(0, n, p)
probabilidad_x_3 = binom.pmf(3, n, p)
probabilidad_x_4 = binom.pmf(4, n, p)

print("La probabilidad de que ninguno sea defectuoso es:", probabilidad_x_0)
print("La probabilidad de que tres sean defectuosos es:", probabilidad_x_3)
print("La probabilidad de que cuatro sean defectuosos es:", probabilidad_x_4)
```
