```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from scipy.stats import pearsonr
import pandas as pd

#Datos de entrada
grupo1=pd.read_csv("/content/grupo1.csv")
horas_estudio1=grupo1[["HorasEstudio"]]
calificacion1=grupo1["Calificacion"]

#Crear y entrenar el modelo de regresión lineal
modelo=LinearRegression()
modelo.fit(horas_estudio1, calificacion1)


#Coeficientes de la regresión
print("Coeficiente de la pediente (m): ",modelo.coef_[0])
print("Intercepcion de la recta (b): ", modelo.intercept_)
```

```python
grupo1=pd.read_csv("/content/grupo1.csv")
horas_estudio1=grupo1["HorasEstudio"]
calificacion1=grupo1["Calificacion"]

plt.figure(figsize=(10,6))

sns.set(style="whitegrid")
sns.regplot(x=horas_estudio1, y=calificacion1, scatter=True, color='green')
plt.title('Diagrama de dispersión')
plt.xlabel('Horas de estudio')
plt.ylabel('Calificaciones')
plt.grid(True)
plt.show()
```

```python
#Datos de entrada
grupo1=pd.read_csv("/content/grupo1.csv")
horas_estudio1_data=grupo1[["HorasEstudio"]]
calificacion1_data=grupo1["Calificacion"]

#Crear y entrenar el modelo de regresión lineal
modelo=LinearRegression()
modelo.fit(horas_estudio1_data,calificacion1_data)

#Predecir la calificación para 8 horas de estudio
horas_estudio_prediccion=pd.DataFrame([8], columns=horas_estudio1_data.columns)
calificacion_predicha=modelo.predict(horas_estudio_prediccion)

print("La calificación esperada para 8 horas de estudio es: ", calificacion_predicha[0])
print("Valor de la intercepción", modelo.intercept_)
print("Valor de la pendiente", modelo.coef_)

#Crear el diagrama de dispersión con la línea de regresión
plt.figure(figsize=(6,5))
plt.scatter(horas_estudio1_data,calificacion1_data,color='blue', label='Datos')
plt.plot(horas_estudio1_data, modelo.predict(horas_estudio1_data), color='red', label='Línea de regresión')
plt.scatter(horas_estudio_prediccion, calificacion_predicha, color='green',marker='x', s=100, label='Predicción (8,{0:.2f})'. format(calificacion_predicha[0]))
plt.title('Pronóstico de calificaciones a partir de las horas de estudio')
plt.xlabel('Horas de estudio')
plt.ylabel('Calificación')
plt.legend()
plt.grid(True)
plt.show()
```

```python

grupo1=pd.read_csv("/content/grupo1.csv")
horas_estudio1=grupo1["HorasEstudio"]
calificacion1=grupo1["Calificacion"]


coeficiente_correlacion,p_valor= pearsonr(horas_estudio1, calificacion1)

print("Coeficiente de correlación de Pearson:", coeficiente_correlacion)
print("Valor p:",p_valor)
```

##1.¿Explicar cuál es la interpretación del coeficiente de regresión en terminos de la relación entre el tiempo de estudio y las calificaciones finales?
###El coeficiente de la pendiente indica cuánto cambia la calificación final por cada hora adicional de estudio.

###En este caso, el valor es negativo (-0.106), lo que significa que, según el modelo, a mayor tiempo de estudio, ligeramente disminuyen las calificaciones finales.


##2.¿Cómo se interpreta el coeficiente de correlación en términos de la fuerza y dirección de la relación entre variables?
###El coeficiente de correlación mide la fuerza y dirección de la relación lineal entre dos variables.

###El valor está muy cercano a 0, lo que indica el tiempo de estudio prácticamente no tiene relación lineal con las calificaciones finales en estos datos.


##3.¿El modelo de regresión lineal es estadisticamente significativo?
###El valor p indica si la relación encontrada es estadísticamente significativa.

###Como el valor p es mucho mayor que 0.05, el modelo no es significativo.

###Esto significa que no podemos afirmar con confianza que el tiempo de estudio tenga un efecto real sobre las calificaciones finales.

##4.¿Qué tan confiable es el modelo de regresión para predecir las calificaciones finales de los estudiantes basado en el tiempo de estudio?
###El modelo de regresión no es confiable para predecir las calificaciones finales basándose en el tiempo de estudio.
