```python
asistentes=["Alberto","Luis","Brian","Jorge","Jesús","Miguel","Uriel","Diego","Maximiliano","Felipe"]
n=10
r=4

Pr=mt.factorial(n)/(mt.factorial(n-r))

print("Existen",Pr,"posibles permutaciones sin repetición del total de ",n,"asistentes","en grupos de",r,"en",r)

print("Aquí las permutaciones")
permutaciones_asistentes=list(permutations(asistentes,r))

#Imprime las permutaciones
for permutacion in permutaciones_asistentes:
  print(permutacion)
```
